```python
import numpy as np
import matplotlib.pyplot as plt

def g(x):
    return 1 - x + 0.5*x**2 - (1.0/6.0)*x**3

x = np.linspace(0, 4, 1000)
plt.figure(figsize=(8,5))
plt.plot(x, np.abs(g(x)) - 1)
plt.axhline(0, color='k', linewidth=1)
plt.xlabel("x = lambda h")
plt.ylabel("|g(x)| - 1")
plt.title("RK3 stability boundary on real axis")
plt.grid(True)
plt.show()
```
