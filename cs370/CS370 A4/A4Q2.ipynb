```python
import numpy as np

def dft(x):
    x = np.array(x, dtype=complex)
    N = len(x)
    W = np.exp(2j*np.pi/N)
    F = np.zeros(N, dtype=complex)
    for k in range(N):
        F[k] = sum(x[n] * W**(-n*k) for n in range(N)) / N
    return F

f = np.array([1,4,1,-4,-1,4,-1,-4], dtype=complex)
g = f[::2]
h = f[1::2]
G = dft(g)
H = dft(h)
F = dft(f)

print("g =", g)
print("h =", h)
print("G =", np.round(G, 10))
print("H =", np.round(H, 10))
print("F =", np.round(F, 10))
```
