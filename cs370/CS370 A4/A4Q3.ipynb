```python
import numpy as np

def dft(x):
    x = np.array(x, dtype=complex)
    N = len(x)
    W = np.exp(2j*np.pi/N)
    F = np.zeros(N, dtype=complex)
    for k in range(N):
        F[k] = sum(x[n] * W**(-n*k) for n in range(N)) / N
    return F

# Example real even vector: f_n = f_{N-n}
f = np.array([3,2,1,0,1,2], dtype=float)
F = dft(f)

print("f =", f)
print("F =", np.round(F, 10))
print("Imag parts:", np.round(F.imag, 10))
N = len(F)
print("Check F[k]=F[N-k]:")
for k in range(N):
    print(k, np.round(F[k],10), np.round(F[(N-k)%N],10))
```
