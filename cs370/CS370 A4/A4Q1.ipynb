```python
import numpy as np

def dft(x):
    x = np.array(x, dtype=complex)
    N = len(x)
    W = np.exp(2j*np.pi/N)
    F = np.zeros(N, dtype=complex)
    for k in range(N):
        F[k] = sum(x[n] * W**(-n*k) for n in range(N)) / N
    return F
```

# Q1(a) verify with N=8

```python
N = 8
n = np.arange(N)
f = np.cos(4*np.pi*n/N) + 1j*np.sin(4*np.pi*n/N)
F = dft(f)
print("Q1(a), N=8:", np.round(F, 10))
```

# Q1(b) verify with N=8

```python
N = 8
f = np.array([0]*(N//2) + [1]*(N//2), dtype=complex)
F = dft(f)
print("Q1(b), N=8:", np.round(F, 10))
print("Even indices F2,F4,F6:", F[2], F[4], F[6])
```
